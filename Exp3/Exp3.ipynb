```python
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = sns.load_dataset("iris")

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Correct column names
columns = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

for ax, col in zip(axes.flat, columns):
    sns.boxplot(x=df[col], ax=ax)
    ax.set_title(f"Box Plot of {col.replace('_', ' ').title()}")

plt.tight_layout()
plt.show()
```

```python
# Plot 3: Multiple Box Plot- Petal Length by Species
sns.boxplot(data=df, x='Species', y='PetalLengthCm')
plt.title('Petal Length by Species')
plt.show()
```

```python
# Plot 3: Multiple Box Plot- Sepal Width by Species
sns.boxplot(data=df, x='Species', y='SepalWidthCm')
plt.title('Sepal Width by Species')
plt.show()
```

```python
ax = sns.countplot(
    data=df,
    x='Species',
    hue='Species',
    palette='Set2',
    legend=False
)

# Display count above each bar
for container in ax.containers:
    ax.bar_label(container)

plt.title('Distribution of Iris Species')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
iris = pd.read_csv("/kaggle/input/iris/Iris.csv")
df = pd.DataFrame(iris)

# Plot 7: Histogram
# Step 18: Histogram (Histplot)

sns.histplot(data=df, x='SepalLengthCm', hue='Species', kde=True)

# histplot shows the distribution (frequency) of a single numerical feature.
# hue='Species' separates the distributions by flower species.
# kde=True adds a smooth density curve to better visualize the distribution.

plt.show()

# Helps identify:
# - Distribution of values (normal, skewed, etc.)
# - Frequency of observations
# - Overlap between different species
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
iris = sns.load_dataset('iris')

# Plot histogram for a feature (e.g., sepal_length)
plt.figure(figsize=(6,4))
plt.hist(iris['sepal_length'], bins=10, color='skyblue', edgecolor='black')

# In a histogram, the "bin value" (or simply "bins")
# refers to how your data is grouped into intervals (ranges).
# Instead of plotting every individual data point, a
# histogram divides the data range into equal intervals called bins.

# Add labels and title
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')

plt.show()

# Helps identify:
# - Distribution of the data
# - Frequency of values
# - Presence of skewness or outliers
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
iris = pd.read_csv("/kaggle/input/iris/Iris.csv")
df = pd.DataFrame(iris)

# Plot 8: KDE Plot
# Step 19: Kernel Density Estimation (KDE) Plot

plt.figure(figsize=(6,4))

sns.kdeplot(data=df, x='PetalLengthCm', fill=True, color='purple')

# A KDE plot shows the probability density (distribution)
# of a numerical feature using a smooth curve.
# Unlike a histogram, it does not use bins.

# Add labels and title
plt.title('KDE Plot of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Density')

plt.show()

# Helps identify:
# - Distribution of the data
# - Peaks (most common values)
# - Skewness and spread
# - Multiple modes (if present)
```
